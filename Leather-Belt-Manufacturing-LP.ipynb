```python
import pyomo.environ as pyo
from pyomo.opt import SolverFactory
```

**Problem Statement:**

Leather Limited manufactures two types of belts: the deluxe model and the regular model. Each type requires $1$ sq yd of leather. A regular belt requires 1 hour of skilled labor, and a deluxe belt requires $2$ hours. Each week, $40$ sq yd of leather and 60 hours of skilled labor are available. Each regular belt contributes \$3 to profit and each deluxe belt, \$4.


**Formulation**

**Decision Variables**

* $x_1$: Number of regular belts produced per week
* $x_2$: Number of deluxe belts produced per week

---

**Objective Function**

Maximize weekly profit ($Z$):

$$\max Z = 3x_1 + 4x_2$$

---

**Constraints**

* **Leather Availability:**
$$x_1 + x_2 \le 40$$


* **Skilled Labor Hours:**
$$x_1 + 2x_2 \le 60$$


* **Non-negativity Constraints:**
$$x_1 \ge 0, \quad x_2 \ge 0$$

```python
model = pyo.ConcreteModel() # create a model

# Decision Variables
model.x1 = pyo.Var(within=pyo.NonNegativeReals) # The variables in continuous and not integer
x1 = model.x1
model.x2 = pyo.Var(within=pyo.NonNegativeReals) # regular belt
x2 = model.x2

# Objective function
model.obj = pyo.Objective(expr = 3*x1 + 4*x2, sense = pyo.maximize)

# Constraint
model.leather_avail = pyo.Constraint(expr = x1 + x2 <= 40)
model.labor_avail = pyo.Constraint(expr = x1 + 2*x2 <= 60)

# Solve
opt = SolverFactory('glpk')
results = opt.solve(model)

print('maximum profit is =', model.obj())
print('Number of delux belt produced (x1) =', x1.value)
print('Number of regular belt produced (x2)=', x2.value)
```
